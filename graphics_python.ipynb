## Shifted sine
$f(x) = a \cdot \sin(bx + c) + d $

```python
params = [1,1,0,0]

params_2 = [2,1,0,0]
params_3 = [1,2,0,0]
params_4 = [1,1,1,0]
params_5 = [1,1,0,1]

def sine_shift(x,arr):
    return arr[0]*np.sin(arr[1]*x+arr[2])+arr[3]

def param_string(arr):
    return f"a = {arr[0]}, b = {arr[1]}, c = {arr[2]}, d = {arr[3]}"

all_params = [params_2,params_3,params_4,params_5]
```

```python
i=3
plt.figure(figsize=(6,4))
plt.plot(x, sine_shift(x,params), label=param_string(params))
plt.plot(x, sine_shift(x,all_params[i]), label=param_string(all_params[i]))
plt.xlabel("$\\alpha$")
plt.ylim(-2.5,2.5)
plt.xticks(np.array([-2, -1.5,-1, -0.5, 0, 0.5,1, 1.5,2])*np.pi, ["$-2\\pi$",
    "$-\\frac{3\\pi}{2}$","$-\\pi$", "$-\\frac{\\pi}{2}$", "0", 
    "$\\frac{\\pi}{2}$", "$\\pi$", "$\\frac{3\\pi}{2}$","$2\\pi$"])
plt.grid()
plt.legend()
plt.savefig(f"Bilder/shifted_sine_{i}.jpeg", dpi=300)
plt.show()
```

## Graphics for Log and exp

```python
x = np.linspace(-1,2, 100)

exp_1 = np.power(2,x)
exp_2 = np.exp(x)
exp_3 = np.power(4,x)
```

```python
i=3
plt.figure(figsize=(6,4))
plt.plot(x, exp_1, label="$f(x) = 2^x$")
plt.plot(x, exp_2, label="$f(x) = e^x = \exp(x)$ ")
plt.plot(x, exp_3, label="$f(x) = 4^x$")
plt.xlabel("$x$")
plt.ylabel("$y$")
#plt.ylim(-2.5,2.5)
plt.grid()
plt.legend()
plt.savefig(f"Bilder/exp.jpeg", dpi=300)
plt.show()
```

```python
x = np.linspace(-1,2, 100)

exp_1 = np.power(0.5,x)
exp_2 = np.exp(-x)
exp_3 = np.power(0.25,x)

plt.figure(figsize=(6,4))
plt.plot(x, exp_1, label="$f(x) = (0.5)^x$")
plt.plot(x, exp_2, label="$f(x) = (1/e)^x = \exp(-x)$ ")
plt.plot(x, exp_3, label="$f(x) = (0.25)^x$")
plt.xlabel("$x$")
plt.ylabel("$y$")
#plt.ylim(-2.5,2.5)
plt.grid()
plt.legend()
plt.savefig(f"Bilder/exp_decreasing.jpeg", dpi=300)
plt.show()
```
